```python

import matplotlib.pyplot as plt

# Tirada Revenue True iyo False
revenue_counts = data["Revenue"].value_counts().reindex(
    [True, False], fill_value=0
)

print("Revenue counts:")
print(revenue_counts)

# Pie Chart
plt.figure(figsize=(7, 7))

plt.pie(
    revenue_counts,
    labels=["True (Purchase)", "False (No Purchase)"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Online Shoppers Revenue")
plt.show()
```

```python
monthly_revenue = data.groupby("Month")["Revenue"].sum()

month_order = [
    "Feb", "Mar", "May", "June", "Jul",
    "Aug", "Sep", "Oct", "Nov", "Dec"
]

monthly_revenue = monthly_revenue.reindex(month_order).fillna(0)
print(monthly_revenue)
monthly_revenue.plot(kind="bar")

plt.title("Revenue by Month")
plt.xlabel("Month")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
